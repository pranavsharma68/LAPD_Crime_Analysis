### 4) Data Cleaning

```python
cols_nulls = df.isnull().sum()
```

```python
cols_nulls
```

```python
columns = df.columns
```

```python
for i in columns:
    col_nulls2 = df[i].isnull().mean()
    if col_nulls2 > 0.3:
        print(f"{i} has {col_nulls2:.2%} missing values")
```

```python
for i in columns:
    if(df[i].isnull().mean() > 0.3):
        df.drop(i, axis=1, inplace=True)
```

```python
df['time_str'] = df['TIME OCC'].astype(str).str.zfill(4)

df['Time Occured'] = pd.to_datetime(df['time_str'], format='%H%M').dt.strftime('%H:%M')

print(df.head)
```

```python
# Invalid time check
invalid_times = []

for time in df['TIME OCC']:
    if pd.isnull(time):
        continue
    hour = int(time) // 100
    minute = int(time) % 100
    if hour > 23 or minute > 59:
        invalid_times.append(time)

print("Invalid TIME OCC values found:", len(invalid_times))
print("Examples:", invalid_times[:10])

#doubt
```
